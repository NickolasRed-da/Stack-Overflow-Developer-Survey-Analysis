```python
# === Завдання 2 ===

import pandas as pd

df = pd.read_csv('survey_results_public.csv', low_memory=False)
schema = pd.read_csv('survey_results_schema.csv')

# Список питань зі схеми
schema_questions = set(schema['qname'].unique())

print(f'Кількість унікальних питань у схемі (qname): {len(schema_questions)}')
print(f'Перші 10 питань: {list(schema_questions)[:10]}')

# Перевірка наявність відповідей на питання у кожного респондента
df_columns = set(df.columns)
common_columns = schema_questions & df_columns

print(f'\nКолонок у датасеті: {len(df_columns)}')
print(f'Питань у схемі: {len(schema_questions)}')
print(f'Спільних колонок (інтерсекція): {len(common_columns)}')

# Підрахунок респондентів без пропусків
df_filtered = df[list(common_columns)]
complete_respondents = df_filtered.dropna().shape[0]

print(f'\nКількість респондентів, які відповіли на всі запитання: {complete_respondents}')
```

```python
# === Завдання 3 ===

# Колонка досвіду роботи (пропущені значення ігноруються)
work_exp = df['WorkExp'].dropna()

mean_val   = work_exp.mean()
median_val = work_exp.median()
mode_val   = work_exp.mode()[0]

print("=== Статистичний аналіз досвіду роботи (WorkExp) ===\n")
print(f"Кількість відповідей : {len(work_exp)}")
print(f"Пропущених значень   : {df['WorkExp'].isna().sum()}\n")
print(f"Середнє (mean)       : {mean_val:.2f} років")
print(f"Медіана (median)     : {median_val:.2f} років")
print(f"Мода (mode)          : {mode_val:.2f} років")
```

```python
# === Завдання 4 ===

df = pd.read_csv('survey_results_public.csv', low_memory=False)

remote_count = (df['RemoteWork'] == 'Remote').sum()

print("=" * 45)
print("   Аналіз віддаленої роботи (RemoteWork)")
print("=" * 45)
print(f"\n Кількість повністю віддалених респондентів: {remote_count}")
print(f"\nРозбивка по всіх форматах роботи:")
print("-" * 45)
print(df['RemoteWork'].value_counts().to_string())
print("-" * 45)
print(f"Всього відповіли на питання: {df['RemoteWork'].notna().sum()}")
print(f"Пропущених значень: {df['RemoteWork'].isna().sum()}")
```

```python
# === Завдання 5 ===

import pandas as pd

df = pd.read_csv('survey_results_public.csv', low_memory=False)

total = len(df)
python_users = df['LanguageHaveWorkedWith'].str.contains('Python', na=False).sum()
percentage = (python_users / total) * 100

print("=" * 40)
print("  Популярність Python серед респондентів")
print("=" * 40)
print(f"\n Використовують Python:  {python_users} осіб")
print(f" Всього респондентів:    {total} осіб")
print(f"\n Відсоток Python-розробників: {percentage:.1f}%")
print("=" * 40)
```

```python
# === Завдання 7 ===

import pandas as pd

df = pd.read_csv('survey_results_public.csv', low_memory=False)

python_devs = df[df['LanguageHaveWorkedWith'].str.contains('Python', na=False)]

result = (
    python_devs
    .groupby('Country')['ConvertedCompYearly']
    .agg(
        Середня_компенсація='mean',
        Медіанна_компенсація='median',
        Кількість_розробників='count'
    )
    .round(2)
    .sort_values('Медіанна_компенсація', ascending=False)
    .reset_index()
)

result.columns = ['Країна', 'Середня компенсація ($)', 'Медіанна компенсація ($)', 'Кількість розробників']

pd.set_option('display.float_format', lambda x: f'{x:,.2f}')
pd.set_option('display.max_rows', None)
pd.set_option('display.width', 120)

print("=" * 75)
print("   Географічний аналіз компенсації Python-розробників")
print("=" * 75)
print(f"\nВідфільтровано Python-розробників: {len(python_devs):,}")
print(f"З даними про компенсацію:          {python_devs['ConvertedCompYearly'].notna().sum():,}")
print(f"Країн у вибірці:                   {result['Країна'].nunique()}\n")
print(result.to_string(index=False))
print("=" * 75)
```

```python
# === Завдання 8 ===

top5 = (
    df[df['ConvertedCompYearly'].notna()]
    .sort_values('ConvertedCompYearly', ascending=False)
    .head(5)
    [['ConvertedCompYearly', 'EdLevel', 'Country']]
    .reset_index(drop=True)
)

top5.index += 1
top5.columns = ['Компенсація ($)', 'Рівень освіти', 'Країна']
top5['Компенсація ($)'] = top5['Компенсація ($)'].apply(lambda x: f'{x:,.0f}')

print("=" * 75)
print("   Топ-5 найбільш оплачуваних спеціалістів — рівні освіти")
print("=" * 75)
print(top5.to_string())
print("=" * 75)
```

```python
# === Завдання 9 ===

df['uses_python'] = df['LanguageHaveWorkedWith'].str.contains('Python', na=False)

result = (
    df.groupby('Age', dropna=True)
    .agg(
        Всього=('uses_python', 'count'),
        Python_розробників=('uses_python', 'sum')
    )
    .assign(Відсоток_Python=lambda x: (x['Python_розробників'] / x['Всього'] * 100).round(1))
    .sort_values('Відсоток_Python', ascending=False)
    .reset_index()
)

result.columns = ['Вікова категорія', 'Всього респондентів', 'Python-розробників', 'Відсоток Python (%)']

print("=" * 70)
print("   Популярність Python по вікових категоріях")
print("=" * 70)
print(result.to_string(index=False))
print("=" * 70)
print(f"\nПропущених значень у Age: {df['Age'].isna().sum()}")
```

```python
# === Завдання 10 ===

import pandas as pd

df = pd.read_csv('survey_results_public.csv', low_memory=False)

p75 = df['ConvertedCompYearly'].quantile(0.75)

filtered = df[
    (df['ConvertedCompYearly'] > p75) &
    (df['RemoteWork'] == 'Remote')
]

industry_counts = (
    filtered['Industry']
    .value_counts()
    .reset_index()
)
industry_counts.columns = ['Індустрія', 'Кількість']
industry_counts['Відсоток (%)'] = (industry_counts['Кількість'] / len(filtered) * 100).round(1)

print("=" * 65)
print("  Індустрії високооплачуваних віддалених працівників")
print("=" * 65)
print(f"\n 75-й перцентиль компенсації: ${p75:,.0f}/рік")
print(f" Відфільтровано респондентів:  {len(filtered):,}")
print(f"   (компенсація > p75 + повністю віддалена робота)\n")
print(industry_counts.to_string(index=False))
print("=" * 65)
print(f"\nПропущених значень у Industry: {filtered['Industry'].isna().sum()}")
```
